```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
adult19 = pd.read_csv('adult19.csv')
adult19[['year']] = '2019'
adult20 = pd.read_csv('adult20.csv')
adult20[['year']] = '2020'
adult21 = pd.read_csv('adult21.csv')
adult21[['year']] = '2021'
```

```python
# columns19 = adult19[['PHQCAT_A', 'GADCAT_A', 'PAIFRQ3M_A','year']]
# columns20 = adult20[['DEPLEVEL_A','ANXLEVEL_A','PAIFRQ3M_A','year']]
# columns21 = adult21[['DEPLEVEL_A','ANXLEVEL_A','PAIFRQ3M_A','year']]
columns19 = adult19[['PHQCAT_A',  'PAIFRQ3M_A','year']]
columns20 = adult20[['DEPLEVEL_A','PAIFRQ3M_A','year']]
columns21 = adult21[['DEPLEVEL_A','PAIFRQ3M_A','year']]
```

```python
for col in columns19:
  columns19[col].replace([7,8,9], np.NaN ,inplace=True)
for col in columns20:
  columns20[col].replace([7,8,9], np.NaN ,inplace=True)
for col in columns21:
  columns21[col].replace([7,8,9], np.NaN ,inplace=True)
```

```python
columns19 = columns19.rename(columns={'PHQCAT_A':'DEP','GADCAT_A':'ANX', 'PAIFRQ3M_A':'PAIN'})
columns20 = columns20.rename(columns={'DEPLEVEL_A':'DEP','ANXLEVEL_A':'ANX', 'PAIFRQ3M_A':'PAIN'})
columns21 = columns21.rename(columns={'DEPLEVEL_A':'DEP','ANXLEVEL_A':'ANX', 'PAIFRQ3M_A':'PAIN'})
```

```python
df = pd.concat([columns19,columns20,columns21])
df.dropna(inplace=True)
```

```python
df.isna().sum()
```

```python
for col in df:
  df[col].replace([1,2], 0 ,inplace=True)
  df[col].replace([3,4], 1 ,inplace=True)
```

```python
df.ANX.value_counts()
```

```python
df['A\D'] = df['DEP'].astype(bool) | df['ANX'].astype(bool)
```

```python
df.dropna(inplace=True, subset=['A\D','PAIN'])
```
